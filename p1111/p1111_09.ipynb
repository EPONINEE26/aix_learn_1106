```python
# 그래프관련 import 
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False

# 프로그램 
import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier # 분류 클래스
from sklearn.neighbors import KNeighborsRegressor # 예측 클래스
from sklearn.linear_model import LinearRegression # 예측 클래스
from sklearn.preprocessing import PolynomialFeatures # 다항특성 자동생성 라이브러리
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import mean_absolute_error # 평균절대오차 오차범위 클래스
from sklearn.preprocessing import StandardScaler  # 정규화 클래스
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('data/MultipleLinearRegressionData.csv')
df.head(5)
```

```python
df[['hour','absent']]
data = df.iloc[:,:-2].values
label = df.iloc[:,-1].values
label
```

```python
# 학습데이터,테스트데이터 분리
train_data,test_data,train_label,test_label = train_test_split(total_data,label,random_state=42)
```

```python
lr = LinearRegression()
lr.fit(train_data,train_label)
score = lr.score(train_data,train_label)
score2 = lr.score(test_data,test_label)
print(score)
print(score2)
```

```python
data = df.iloc[:,:-2].values
label = df.iloc[:,-1].values
label
```

```python
# 2. 훈련데이터, 테스트데이터 분리
train_data, test_data, train_label, test_label = train_test_split(data, label, random_state=42, test_size=0.2)
```

```python
# 3. 다항 특성 자동 생성
poly = PolynomialFeatures(degree=5, include_bias=False)
poly.fit(train_data)
train_poly = poly.transform(train_data)
test_poly = poly.transform(test_data)
```

```python
print(poly.get_feature_names_out())
```

```python
# 4. 표준화
ss = StandardScaler()
ss.fit(train_poly)
train_scaled = ss.transform(train_poly)
test_scaled = ss.transform(test_poly)
```

```python
# 5. 릿지 회귀 - 규제 강도 탐색
train_score = []
test_score = []
alpha_list = [0.001, 0.01, 0.1, 1, 10, 100]
for alpha in alpha_list:
    ridge = Ridge(alpha=alpha)
    ridge.fit(train_scaled, train_label)
    train_score.append(ridge.score(train_scaled, train_label))
    test_score.append(ridge.score(test_scaled, test_label))
```

```python
# 6. alpha별 성능 그래프
plt.plot(np.log10(alpha_list), train_score, marker='o')
plt.plot(np.log10(alpha_list), test_score, marker='o')
plt.xlabel('log10(alpha)')
plt.ylabel('R2 Score')
plt.legend(['Train', 'Test'])
plt.grid()
plt.show()
```

```python
ridge = Ridge(alpha=1)
ridge.fit(train_scaled, train_label)
print('[최종 모델 성능]')
print('훈련 정확도:', ridge.score(train_scaled, train_label))
print('테스트 정확도:', ridge.score(test_scaled, test_label))
```

```python
# 9. 예측 및 오차 확인
result = ridge.predict(test_scaled)
mae = mean_absolute_error(test_label, result)
print('평균 절대 오차(MAE):', mae)
```
