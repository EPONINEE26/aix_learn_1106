```python
from requests import get
import pandas as pd
```

```python
url = "http://kobis.or.kr/kobisopenapi/webservice/rest/movie/searchMovieList.json?key=ec226160f0741bf033b13c3551dbaa6a"

header = {}

response = get(url, headers=header)
data = response.json()

movie_list = data["movieListResult"]["movieList"][:10]

df = pd.DataFrame(movie_list)
df.to_csv("movie_list_10.csv", index=False, encoding="utf-8-sig")

print(df)
```

```python
print("서버에서 받은 영화 개수 :", len(data["movieListResult"]["movieList"]))
print("선택한 영화 개수 :", len(movie_list))
```

```python
from requests import get
import pandas as pd

url = "http://kobis.or.kr/kobisopenapi/webservice/rest/movie/searchMovieList.json?key=ec226160f0741bf033b13c3551dbaa6a&itemPerPage=100"

header = {}

response = get(url, headers=header)
data = response.json()

movie_list = data["movieListResult"]["movieList"]

df = pd.DataFrame(movie_list)

df.to_csv("movie_list_100.csv", index=False, encoding="utf-8-sig")

print("영화 개수 :", len(df))
```

```python
print("서버에서 받은 영화 개수 :", len(data["movieListResult"]["movieList"]))
print("선택한 영화 개수 :", len(movie_list))
```
