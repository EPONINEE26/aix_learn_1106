```python
df = pd.read_csv('data/iris.csv', index_col='caseno')
df.head(5)
```

```python
# data = df[['SepalLength',	'SepalWidth',	'PetalLength',	'PetalWidth']].values
data = df[df.columns[1:-1]].values
label = df['Species'].values

train_data, test_data, train_label, test_label = train_test_split(data,label,test_size=0.2,random_state=42)
```

```python
# rf = RandomForestClassifier(n_jobs=-1,random_state=42)
rf = RandomForestClassifier(n_estimators=100)
rf.fit(train_data,train_label)

print('train정확도:', rf.score(train_data,train_label))
print('test정확도:', rf.score(test_data,test_label))
```

```python
rf = RandomForestClassifier(n_estimators=100)
rf.fit(train_data,train_label)

print('train정확도:', rf.score(train_data,train_label))
print('test정확도:', rf.score(test_data,test_label))

# score = cross_validate(rf, train_data,train_label,return_train_score=True,n_jobs=-1)
# print(score['train_score'])
# print(score['test_score'])
```
