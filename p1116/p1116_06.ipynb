```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False
import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso,LogisticRegression
from sklearn.metrics import mean_absolute_error   
from sklearn.preprocessing import PolynomialFeatures,StandardScaler
from sklearn.model_selection import train_test_split
from scipy.special import expit, softmax    
from sklearn.linear_model import SGDClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from sklearn.model_selection import cross_validate,StratifiedKFold,GridSearchCV
import statsmodels.api as sm
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder
```

```python
# 무게를 에측하는 알고리즘 구현하시오
# 길이, 높이, 넓이를 입력받아 무게를 예측하시오.
# length, height, width -> weight 예측 
fish = pd.read_csv('data/perch_full.csv')
fish.head(5)
```

```python
perch_weight = [5.9, 32.0, 40.0, 51.5, 70.0, 100.0, 78.0, 80.0, 85.0, 85.0,
     110.0, 115.0, 125.0, 130.0, 120.0, 120.0, 130.0, 135.0, 110.0,
     130.0, 150.0, 145.0, 150.0, 170.0, 225.0, 145.0, 188.0, 180.0,
     197.0, 218.0, 300.0, 260.0, 265.0, 250.0, 250.0, 300.0, 320.0,
     514.0, 556.0, 840.0, 685.0, 700.0, 700.0, 690.0, 900.0, 650.0,
     820.0, 850.0, 900.0, 1015.0, 820.0, 1100.0, 1000.0, 1100.0,
     1000.0, 1000.0]
```

```python
train_data,test_data,train_label,test_label = train_test_split(fish,perch_weight)
```

```python
# 다중회귀 특성 poly 추가
poly = PolynomialFeatures(degree=5,include_bias=False)
poly.fit(train_data)
train_poly = poly.transform(train_data) # 특성값 추가
test_poly = poly.transform(test_data) 

# 정규화작업 
ss = StandardScaler()
ss.fit(train_poly)
train_scaled = ss.transform(train_poly)
test_scaled = ss.transform(test_poly)

# 리지모델 
ridge = Ridge()

# 훈련 
ridge.fit(train_scaled,train_label)

# 예측 
# 43.0,	12.51,	7.42 -> 입력받아 무게를 예측
# 21.3,	6.38,	3.53
# input('길이,높이,넓이를 입력하세요. 예)40,11.73,7.22')

in_data = input('길이,높이,넓이를 입력하세요. 예)40,11.73,7.22')
in_data = in_data.split(',')
result = list(map(float, in_data))

# 다중회귀 - 특성추가 
result_poly = poly.transform([result])

# 정규화 작업 
result_scaled = ss.transform(result_poly)

# 예측 
predict = ridge.predict(result_scaled)
print(f'예측무게 : ({predict[0]:.2f}g)')

# 정확도 
print('train정확도:', ridge.score(train_scaled,train_label))
print('test정확도:', ridge.score(test_scaled,test_label))
```
