```python
import joblib
model = joblib.load('model.joblib')
```

```python
# 예측 
in_data = input('길이,높이,넓이를 입력하세요. 예)40,11.73,7.22')
# print(in_data)
in_data = in_data.split(',')
result = list(map(float, in_data))
# print(result)

# 예측
print('입력데이터:', result)
result = model.predict([result])
print('예측무게: {result[0]:.2f}g')
```
