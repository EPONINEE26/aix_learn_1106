```python
# 검증세트 : 양이 적을 경우 효율적임 
```

```python
wine = pd.read_csv('data/wine.csv')
wine.head(5)
```

```python
data = wine[wine.columns[:-1]].values
label = wine['class'].values
label[:5]
```

```python
train_data, test_data, train_label, test_label = train_test_split(data, label, test_size=0.2, random_state=42)
```

```python
sub_data, sub2_data, sub_label, sub2_label = train_test_split(train_data, train_label, test_size=0.2, random_state=42)
```

```python
# max_depth : 과대적합일 때 max_depth줄이면 과대적합 해소 
# dt = DecisionTreeClassifier(max_depth=3,random_state=42)
dt = DecisionTreeClassifier(random_state=42) 
dt.fit(train_data,train_label)
print()
```

```python
print('train 정확도:', dt.score(sub_data,sub_label))
print('test 정확도:', dt.score(sub2_data, sub2_label))
```

```python
print('sub 정확도:', dt.score(sub_data,sub_label))
print('sub2 정확도:', dt.score(sub2_data, sub2_label))
```

```python
print('test 정확도:', dt.score(test_data,test_label))
```
