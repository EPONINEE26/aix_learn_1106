```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False
import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso,LogisticRegression
from sklearn.metrics import mean_absolute_error   
from sklearn.preprocessing import PolynomialFeatures,StandardScaler
from sklearn.model_selection import train_test_split
from scipy.special import expit, softmax    
from sklearn.linear_model import SGDClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from sklearn.model_selection import cross_validate
from sklearn.model_selection import StratifiedKFold
```

```python
wine = pd.read_csv('data/wine.csv')
wine.head(5)
```

```python
data = wine[wine.columns[:-1]].values
label = wine['class'].values
label[:5]
```

```python
train_data, test_data, train_label, test_label = train_test_split(data, label, test_size=0.2, random_state=42)
```

```python
dt = DecisionTreeClassifier(max_depth=3, random_state=42) 
# fit 부분을 5번 반복함 (기본 : 5번)
# cv = 5 : 훈련 데이터를 5등분 하라는 의미 -> 5번 실행 
# cv : 등분하는 개수 (기본값 : 5)
# 훈련 데이터를 3등분 -> 3번 실행 
# default : test_score 만 출력, return_train_score=True : train_score, test_score 모두 출략
score = cross_validate(dt, train_data, train_label,cv=5,return_train_score=True)
# fit_time : 훈련시간, score_time : score 생성시간, test_score : test정확도, train_score : train정확도 
print(score)
```

```python
print("train 평균정확도 :", np.mean(score['train_score']))
print("test 평균정확도 :", np.mean(score['test_score']))
```
