```python
# 데이터 전처리
# 학습데이터, 테스트데이터 분리
train_data, test_data, train_label, test_label = train_test_split(df,perch_weight,random_state=42)

# 데이터 학습 
lr = LinearRegression()
lr.fit(train_data,train_label)

# 데이터 예측
predict = lr.predict(test_data)

# 정확도 
print('train :', lr.score(train_data,train_label))
print('test :', lr.score(test_data,test_label)) 
```

```python
# 특성을 추가해서 정확도를 높이는 작업 
poly = PolynomialFeatures(degree=2,include_bias=False)
poly.fit(train_data)
train_poly = poly.transform(train_data)
test_poly = poly.transform(test_data)
```

```python
print(train_data.shape)
print(train_poly.shape)
poly.get_feature_names_out()
```

```python
# 규제 - 정규화 작업, 규제알고리즘 사용 

ss = StandardScaler()
ss.fit(train_poly)
train_scaled = ss.transform(train_poly)
test_scaled = ss.transform(test_poly)
```

```python
# 규제 알고리즘 
# test 정확도가 낮으면 규제 degree를 높여서 정확도를 높일 수 있음
ridge = Ridge()
ridge.fit(train_scaled,train_label)
print('train 정확도:', ridge.score(train_scaled,train_label))
print('test 정확도:', ridge.score(test_scaled,test_label))
```
