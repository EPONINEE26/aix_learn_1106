```python
data = fish[['Weight', 'Length', 'Diagonal', 'Height', 'Width']].values
label = fish['Species'].values
```

```python

train_data, test_data, train_label, test_label = train_test_split(data,label)
```

```python
ss = StandardScaler()
ss.fit(train_data)
train_scaled = ss.transform(train_data)
test_scaled = ss.transform(test_data)
```

```python
# 2개만 분류 - 빙어, 도미 
# KNeighborsClassifier - 다중 분류
 
clf = KNeighborsClassifier(n_neighbors=5) # 근접한 정보를 가지고 분류
# 세미콜론(;)을 붙이는 이유는 Jupyter Notebook(또는 Colab)에서 셀 마지막 줄의 실행 결과를 화면에 출력하지 않기 위해서/
# 컴퓨터에 파이썬 패키지들이 저장되는 site-packages 폴더 안의 sklearn 디렉토리 바로 밑에 위치하고 있는 사이킷런의 _config.py 파일이 
# 화면에 그림을 그리기 위해 그림의 재료가 되는 외부 자바스크립트 파일(예: jquery.js나 웹 라이브러리 파일) 및 웹 스타일 소스를 외부 서버나 
# 내부 경로에서 읽어와야 하지만, 외부 자바스크립트 파일(스크립트 파일)과 소스를 아예 못 읽어오는 상황이 발생함. 
# 그래서 코드 맨 뒤에 세미콜론 ;을 붙여서 이 소스를 읽어오려는 시도와 출력 자체를 깔끔하게 막아버리는 것
clf.fit(train_scaled, train_label);s
```

```python
print("결과 항목 : ", clf.classes_) # 분류형태가 몇 가지인지 알려줌, 알파벳 순서로 배열
```

```python
label.unique()
```

```python
predict = clf.predict(test_scaled)
print('예측 생선 : ', predict)
```

```python
print('train 정확도 :', clf.score(train_data,train_label))
print('test 정확도 :', clf.score(test_data,test_label))
```

```python
proba = clf.predict_proba(test_scaled[:5]) # 각 튻어의 결정 확률을 보여줌 
proba # 각 확률의 % 를 보여줌 
print("결과 항목 : ", clf.classes_) # (n_neighbors=5) 인 경우 50%, (n_neighbors=3) 인 경우 30%  확률형태로 보여줌, 결과 항목이 꼭 있어야함. 그래야만 각 항목의 % 확인 가능, label특성 항목 - 알파벳 순서로 정렬
print(np.round(proba,decimals=4))
```
