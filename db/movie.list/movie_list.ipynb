```python
from requests import get
import pandas as pd
```

```python
# itemPerPage=10 이 코드가 없는 이유는 기본으로 10개는 제공하기에 사이트만 불러오면 10개는 저장 가능 
url = "http://kobis.or.kr/kobisopenapi/webservice/rest/movie/searchMovieList.json?key=ec226160f0741bf033b13c3551dbaa6a"
header = {}

# movie_data는 서버 응답 객체이고, data는 JSON 변환 결과
movie_data = get(url,headers=header)
data = movie_data.json()

# movies : 영화 목록을 저장하기 위해 직접 만든 변수 이름
# ["movieListResult"] : 전체 JSON에서 영화 목록 조회 결과 선택
# ["movieList"] : 조회 결과 안에 들어 있는 실제 영화 목록 선택
movie_list = data["movieListResult"]["movieList"]
# movies = data["movieListResult"]["movieList"]
# result = data["movieListResult"]["movieList"]
df = pd.DataFrame(movie_list)
# df = df.drop(columns=["directors", "companys"])  # 감독, 제작사 정보 제외
df.to_csv("movie_list_10.csv", index=False, encoding="utf-8-sig")

print(df)

# response = get(url, headers=header)
# data = response.json()

# movie_list = data["movieListResult"]["movieList"][:10]
# df = df.drop(columns=["directors", "companys"])  # 감독, 제작사 정보 제외
# df = pd.DataFrame(movie_list)
# df.to_csv("movie_list_10.csv", index=False, encoding="utf-8-sig")

# print(df) 

```
