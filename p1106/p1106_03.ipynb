```python
from sklearn import svm, metrics
import pandas as pd
```

```python
df = pd.read_csv('data/iris.csv', index_col='caseno')
# df.info()
# df.describe()
# df.shape
# df.columns
df['Species'].unique()
```

```python
# df.columns
df.head()
```

```python
# 1. 데이터 준비 
# - 전체 데이터를 가져옴
# - 정답컬럼 문자를 숫자로 변경 
# - 데이터 분리 : data:학숩, label:정답 분리
# - 훈련데이터, 테스트데이터 분리 (8:2)
# - 빈 공백처리를 어떻게 할지를 정확하게 정해야함 

# 2. 데이터 학습 fit
# 3. 데이터 예측 prefict
# 4. 데이터 검증 score 
# 숫자가 높으면 중요하다고 생각함. 그렇기게 지정 숫자를 잘 선택해야함
```

```python
# 데이터 준비 
# 150개 - str 컬럼을 변경해야함 
# df['Species'].unique()
def nameChange(s):
    if s == 'Iris-sentosa':
        return 0
    elif s == 'Iris-versicolor':
        return 1
    else : 
        return 2 # Iris-virginica
df['Species'] = df['Species'].apply(nameChange)
df.head()
# df['Species'].unique()
# df['Species']
```

```python
# 데이터 분리 - data, label
# x_data, x_test, y_data, y_test
# x_train, x_test, y_train, y_test
# train_data, test_data, train_label, test_label 

# 데이터 학습 - 150 
data = df[['SepalLength','SepalWidth','PetalLength','PetalWidth']]
# 데이터의 정답 
label = df['Species']
data
```

```python
# 150개 데이터에서학습데이터, 테스트데이터를 분리
# 8:2 비율로 학습데이터와 테스트데이터를 분리 
# train_data = df.sample(frac=0.8, random_state=1)
# train_data.shape

# 학습데이터, 테스트데이터 분리 함수 
# train_test_split() 함수 사용
from sklearn.model_selection import train_test_split
train_data, test_data, train_label, test_label = train_test_split(data, label)
```

```python
from sklearn.model_selection import train_test_split
# test_size=0.2 : 20% 를 테스트데이터로 분리 
# stratify=label : lable 의 비율을 유지하면서 뷴리 
# train_data, test_data, train_label, test_label = train_test_split(data, label, test_size=0.2)
# train_data, test_data, train_label, test_label = train_test_split(data, label, test_size=0.2, stratify=label)

train_data, test_data, train_label, test_label = train_test_split(data, label, test_size=0.2, stratify=label, random_state=0)

```

```python
train_data.shape
train_data.head()
```

```python
# 데이터 학습 fit
clf = svm.SVC()
clf.fit(train_data, train_label) # 120개 학습데이터로 학습

result = clf.predict(test_data) # 
```

```python
# 3. 데이터 예측 : prefict
result = clf.predict(test_data) # 30ro 테스트데이터로 예츨 
print("예측결과:", result)
```

```python
# 4. 데이터 검증 : score 
# score = clf.score(test_data, test_label) 
score = clf.score(test_data, test_label)
print("정답율:", score) 
```
