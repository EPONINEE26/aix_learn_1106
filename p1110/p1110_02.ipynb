```python
# 그래프관련 import 
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False

# 프로그램 
import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('data/mushroom.csv', header=None)
df.head(2)
```

```python
df.shape
```

```python
data = pd.DataFrame({
    'color':['red','blue','green','red'] # 컬러의 종류는 3개 # 해당 컬러만 1로 변경 
})
data = pd.get_dummies(data,dtype=int)
print(data)
```

```python
label = df.iloc[:,0]
data = df.iloc[:,1:]
data = pd.get_dummies(data) # 원핫 인코딩 : 0,1 로 변경해줌. 해당 데이터만 1로 변경 
data.shape

```

```python
data = []
label = []
attr_list = []

for idx, row in df.iterrows():
    label.append(row[0])
    row_data = []
    for v in row[1:]: 
        row_data.append(ord(v)) # ord 형태의 아스키코드!!! # 머신러닝은 수치화로 해야만 실행 가능 
    data.append(row_data)
```

```python
label
data
```

```python
# 데이터 전처리 
# 학습 데이터 와 테스트 데이터 분리
train_data, test_data, train_label, test_label = train_test_split(data, label)
    
# 데이터 학습
clf = KNeighborsClassifier()
clf.fit(train_data,train_label)

# 데이터 예측
result = clf.predict(test_data)
print('예측 : ', result)

# 정확도 
score = clf.score(test_data,test_label)
print('정확률 :', score)

# 성능평가 
report = metrics.classification_report(test_label,result)
```

```python
print("리포트 : \n", report)
```
